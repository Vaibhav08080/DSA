# A graph data strucutre is a collection of nodes that have data and are connected to other nodes

```python
nodes = 6
edges = 7
Graph = [(0,1), (0,3), (0,4), (1,2), (1,5), (2,4), (3,4)]
print(nodes, edges, Graph)

adjList = []   # yahi naam consistently use karo
for i in range(nodes):
    adjList.append([])

for edge in Graph:
    x = edge[0]
    y = edge[1]

    adjList[x].append(y)
    adjList[y].append(x)

for i in range(nodes):
    print(i, "->", adjList[i])
```

```python
nodes = 6
edges = 7
Graph = [(0,1), (0,3), (0,4), (1,2), (1,5), (2,4), (3,4)]

# Initialize adjacency matrix with 0
adjMatrix = []
for i in range(nodes):
    adjMatrix.append([0]*nodes)

# Fill edges
for edge in Graph:
    x = edge[0]
    y = edge[1]
    adjMatrix[x][y] = 1
    adjMatrix[y][x] = 1   # undirected graph ke liye symmetric

# Print matrix
for row in adjMatrix:
    print(row)

```

```python
from collections import deque

# Sample Graph (Adjacency List)
graph = {
    0: [1, 2],
    1: [2, 3],
    2: [4],
    3: [4],
    4: []
}

# ✅ DFS (Recursive)
def dfs_recursive(graph, start, visited=None):
    if visited is None:
        visited = set()
    visited.add(start)
    print(start, end=" ")
    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs_recursive(graph, neighbor, visited)

# ✅ DFS (Iterative with Stack)
def dfs_iterative(graph, start):
    visited = set()
    stack = [start]
    while stack:
        node = stack.pop()
        if node not in visited:
            print(node, end=" ")
            visited.add(node)
            # reversed to match recursive order
            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)

# ✅ BFS (Using Queue)
def bfs(graph, start):
    visited = set([start])
    queue = deque([start])
    while queue:
        node = queue.popleft()
        print(node, end=" ")
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

# -----------------------------
# 🔥 Run the functions
print("DFS Recursive Traversal:")
dfs_recursive(graph, 0)
print("\n\nDFS Iterative Traversal:")
dfs_iterative(graph, 0)
print("\n\nBFS Traversal:")
bfs(graph, 0)
```
