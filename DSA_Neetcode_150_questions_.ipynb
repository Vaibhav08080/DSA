# Valid Anagaram - In this question ham phle check karnege ki length equal h ya ni because length equal ni hogi to anagram kese honge then uske baad python ka sorted function use karlenge kyuki usse kisi bhi word ka sorted answer aajata h and then ham usko return karlenge

```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False
        return( sorted(s) == sorted(t) )
```

#Iske andar do solutions h phla approach h ye ki ham phle do iterations kare matlb do loop chlaye agr hamne do loop chalye to n^2 time complexity hojygi isliye ham hashmap ya khlo dictionary in python use karte h phle ek dictionary create karte h and then uske andar diff check karte h exist karta h ya ni agr ni karta to fir append kardete h aur agr karta h to return kardete h

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        prevMap={}
        for i , num in enumerate(nums):
            diff= target- nums[i]
            if diff in prevMap:
                return [ prevMap[diff] , i]
            prevMap[num] = i
```

# Is question me ham har word me charcter ki frequency gin lenge then us frequency ko key manke usse jo bhi word bnta h use map kardenge and then return kardenge

image.png

```python
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        dih= defaultdict(list) # mapping the charcount to list of Anagram
        for s in strs:
            count = [0]*26

            for c in s:
                count[ord(c)-ord("a")]+=1
            dih[tuple(count)].append(s)
        return list(dih.values())
```

# Is question ko solve karne ke majorly tin tarike h but best one is bucket sort in this shit ham count kar rh h len of this array utne tk count kar rha h and then jo number jitni baar h use uski jgh pe dal rhe h h isse hota h ye ki jitne element h utna hi space hoga jisse time complexity n hojati h


o(n) - time complexity and similarly o(n) hi space complexity


```python
class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        count={}
        freq=[[] for i in range(len(nums)+1)]
        for num in nums:
            count[num]= 1+count.get(num, 0)
        for num , cnt in count.items():
            freq[cnt].append(num)
        res=[]
        for i in range(len(freq)-1 , 0 , -1):
            for num in freq[i]:
                res.append(num)
                if len(res)==k:
                    return res
```

# is question me hame str design and decode karni h ek aisa algorithm design karna h jisme hame koi string mile wo encode hojaye aur similar fashion me decode hojaye ab aisa karne ke liye ham ek strng lenge and uhsme  ek chij karenge ki number of character in aword calculate karlenge and tehn ek delimiter yani ek word jaise hash wgrh hogya use use karenge like 4# iska matlba h ki jo bhi word h usme 4 word h aur hash ke baad wale se start karna counting

```python
class Solution:

    def encode(self, strs: List[str]) -> str:
        res = ""
        for s in strs:
            res += str(len(s)) + "#" + s
        return res

    def decode(self, s: str) -> List[str]:
        res = []
        i = 0

        while i < len(s):
            j = i
            while s[j] != '#':
                j += 1
            length = int(s[i:j])
            i = j + 1
            j = i + length
            res.append(s[i:j])
            i = j

        return res
```
