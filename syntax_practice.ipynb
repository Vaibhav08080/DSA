```python
def countdigit(n):
    if n == 0:
        return 0
    return 1 + countdigit(n // 10)

num = int(input("Enter your number: "))
print(countdigit(num))
```

```python
def printreverse(n):
    if n == 0:
        return
    print(n % 10, end="")
    printreverse(n // 10)

a = int(input("Enter a number to reverse: "))
printreverse(a)
```

```python
# reversing a string without slicing
s = input("Enter a string: ")
rev=""
for char in s:
  rev=char+rev
print(rev)

```
